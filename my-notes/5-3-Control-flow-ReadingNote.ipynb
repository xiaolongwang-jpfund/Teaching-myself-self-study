## `if`语句

````python
if expression:
    statements
````

- 如果表达式`expression`的返回值为**真**（`True`)，就执行`if`语句块内部的`statements`；
- 如果表达式`expression`的返回值为**假**（`False`），就执行`if`之后的一个语句。

````python
import random
r = random.randrange(1, 1000)  # 调用 random 模块里的 randrange() 函数，随机生成一个整数，并把它赋值给变量 r 。范围是 1 到 999:包含1，不包含1000.

if r % 2 == 0:
    print(f'{r} is even.')    # 把这行打印出来。前面的 f 表示这是一个 f-string（格式化字符串），花括号 {r} 会被变量 r 当前的值替换。
````

有时，表达式`expression`返回的**值**有多种情况，而且我们要针对不同的情况做相应的事情，可以这么写：

````python
if expression_1:
    statements_for_expression_1_True
elif expression_2:                  # elif 是 else if 的缩写，两者作用相同。
    statements_for_expression_2_True
elif expression_3:
    statements_for_expression_3_True
elif expression_…:
    statements_for_expression_…_True
````

## `for`循环

先来认识一个 Python 函数：**`range()`**：<br>
可以把`range()`理解成为“**整数、等差数列，生成器**”。

### `range()`函数

`range()`是 Python 的内建函数，它的形式是这样的：
- `range(stop)`
  - 当只有一个参数的时候，这个参数被理解为`stop`，生成一个从`0`到`stop-1`的整数数列。
- `range(start, stop[, step])`
  - `start`参数的默认值是`0`。
  - 如果需要给`range()`函数指定一个起点，就要给它传递两个参数`range(start, stop`)。
  - 第三个参数`step`（步长）是可选的，它相当于“等差数列”的“差”。“差”的默认值是`1`。

### `continue`、`break`和`pass`

| 关键词 | 作用 | 循环会怎样 |
| --- | --- | --- |
| continue | 跳过本轮剩余的代码 |进入*当前循环的下一轮*|
| break | 提前结束循环 | 跳出*所在的那一层循环* |
| pass | 什么也不做 | 照常*继续执行后面的代码* |

用质数例子来看：

````python
for n range(2, 100):         # 外层循环：逐个检查 n。n ==2, 3, …, 99
    if n == 2:
        print(n)
        continue             # 跳过本轮剩下的部分，进入下一个 n
    for i in range(2, n):    # 内层循环：试着找 n 的除数
        if n % i == 0:
            break
    else:
        print(n)             # 内层循环没有遇到 break 才执行
````

#### `continue`：跳到外层循环的下一轮

当 n == 2 时，程序先打印 2，然后执行 `continue`。它**跳过本轮中剩下的代码，直接让外层循环`for n range(2, 100):`执行下一个`n`**，也就是 3。

#### `break`：只跳出它所在的内层循环

当 `n == 6` 时，内层循环第一次取到 `i = 2`：

`6 % 2 == 0`

条件成立，于是执行`break`，跳出的是`for i in range(2, n):`这个内层循环。内层`for`的`else`不执行。但外层循环没有结束，会继续检查下一个`n`，也就是 7 。

#### `pass`：占个位置不改变流程

例如：
```python
if n == 2:
    pass
print(n)
```
`pass`不会跳过`print(n)`，所以`print(n)`仍会执行。

`pass`通常用于先搭好代码结构，之后再补上具体内容。

## `while` 循环

### 以输入1000以内的斐波那契数列的程序为例：

```Python
n = 1000                # 保存 n 的上限为1000。后面的循环会在 a 不再小于1000时停止。
a, b = 0, 1。           # 设置 a、b初始值。即 a = 0；b = 1。
while a < n:           # while 每次都先检查条件。只要 a<1000 为真，就执行缩进的两行；条件为假，就跳出循环。
    print(a, end=' ')  # 打印当前这一轮的 a，并在后面加空格、不换行。第一轮开始时 a 是 0，所以先打印 0。
    a, b = b, a+b      # 更新 a 和 b。同时计算并赋值，= 右边先根据更新前的值计算，再把结果放在左边：左边的 a 取旧的 b；左边的 b 取右边的 a+b的和。
print()                #在while循环结束后，执行一次 print(),实现换行。只要 while 循环条件满足，它就不执行。
```

### 以掷骰子游戏程序为例：

```python
…

def bet(dice, wager):
    if dice == 7:
        …
```
与
```python
while True:
    …
```
`def`与`while`是定义工具与使用工具并重复执行的关系。

#### `def bet(dice, wager)`：定义一次下注的处理规则

第6行开始定义`bet`函数。它接收两个值：
- `dice`: 骰子点数
- `wager`：拥护输入的下注方向
函数根据点数和下注方向判断输赢，打印结果，并用`return`返回结果：赢了返回`1`，输了返回`-1`，平局返回`0`。

**定义函数时，这些判断还不会执行**。这一步相当于先准备好一套规则。

#### `while True`：反复进行游戏

第25行开始进入游戏循环。每一轮会显示双方金币、随机生成点数，并询问用户下注。

当输入`b`或`s`时，循环中的这行会调用刚才定义的函数：`result = bet(dice, wager)`。它把本轮的`dice`和`wager`交给`bet`函数。函数运行完成后，`return`的结果会保存进`result`，随后循环代码根据这个结果更新金币和游戏轮数。

#### 两部分怎样配合

可以把程序流程概括成：

先定义`bet`函数 ➡️ 进入`while`游戏循环 ➡️ 每轮产生点数并读取下注 ➡️ 调用`bet(dice, wager)`判断输赢 ➡️ 用返回值更新金币 ➡️ 继续下一轮，或用 `break`结束。

`while`多次调用`bet`，而`bet`每次只处理这一轮下注。输入`q`，或者金币用完时，`break`会结束`while`循环；之后执行循环外的两条`print`，显示总结信息。
