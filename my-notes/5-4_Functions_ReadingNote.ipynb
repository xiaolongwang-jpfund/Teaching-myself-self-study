## 什么是函数

**函数是可以被调用的完整程序**。它具有**输入**、**处理**、**输出**的功能。

函数经常在主程序里被调用，所以它更像是主程序里的一个**子程序**。

了解一个函数，从两个方面入手：
1. 它的**输入**是如何构成的——都有哪些**参数**，如何指定。
2. 它的**输出**是什么——**返回值**是什么？

### `Functions`为什么译成“函数”？

这里的`Functions`译作“函数”，主要是因为它借用了数学里的“函数”这个概念：<br>给它输入，按照某种规则处理，得到输出。

例如，数学里的函数 f(x)=x+1：给它输入3，结果就是4。

所以，初学时可以把`functions`理解为：**一个有名字，可以调用来完成某件事的程序部件**。

读`functions`说明时，先看它接收什么输入、会做什么、返回什么结果。

## 示例：`print()`函数

`print()`函数最基础的作用有两个：
- 把传递给它的值输出到屏幕上。
- 如果不给它传递任何值，那么它就输出一个空行。

### `print()`函数到官方文档说明

[`print()`函数的官方文档](https://docs.python.org/3/builtins/functions.html#print)有很多内容，**必须读懂的部分，就是这一行**：<br>
`print(*object, sep=' ', end='\n', file=sys.stdout, flush=False)`

括号内各参数的含义分别是：
- `*object`：表示`print()`可以接收零个、一个或多个位置参数。可以理解成“**可以传入多个要打印的内容**。
  - 这里的星号`*`是函数说明中的一种标记，意思是“**这里可以有多个值”。
  - `object`只是文档用来代表这些值的名称。
- `sep=' '`表示接收多个参数之后，输出时，**分隔符号默认为空格`' '`**。
- `end='\n'`：在输出行的末尾，**默认是换行符号`'\n'`。
- `file=sys.stdout`：默认的输出对象是`sys.stdout`（即，**用户正在使用的屏幕**）。
- `flush=False`：按照**通常方式处理输出**。这是默认值。

## 一个有意思的拓展

教材中说：
> 原来`print()`函数是可以往文件里写数据的，只要指定`file`这个参数为一个已经打开的文件对象就可以了（真的有很多人完全不知道）。

具体的实现方式，可以看这个示例：

现在`Python`交互环境或`Jupyter Lab` 里运行下面这段完整代码：

```python
with open('print_demo.txt', 'w', encoding='utf-8') as output_file:
    print('你好，文件！', file=output_file)
```

它分成两步：

1. `open(…)`打开名为`print_demo.txt`的文件，得到文件对象`output_file`。
  - `'w'`是写入模式：文件不存在就创建，已经存在就**清空旧内容再写**。
  - `encoding=uft-8`让中文按UTF-8编码写入。

2. `print(…, file=output_file)`把文字写入这个对象，而不是屏幕。`print()`默认会在文字末尾加换行。

`with`会在代码块结束时自动关闭文件，这样数据能妥善写入，也不必另外调用关闭操作。

接下来用下面的代码把文件内容读回来：  

```python
with open('print_demo.txt', 'r', encoding='utf-8') as input_file:
    print(input_file.read())
```

这里`'r'`是只读模式，`input_file.read()`读取文件里的文字；这次`print()`没有指定`file`，所以内容会显示在屏幕上。

注意：写入模式`'w'`会覆盖同名文件。这个例子会在**当前工作目录**创建`print_demo.txt`。

## 函数的参数

在`Python`中，函数的参数有两种：

- **位置参数(Positional Argument)**，在官方文档里常被缩写为`arg`。
- **关键字参数(Keyword Argument)**，在官方文档里常被缩写为`kwarg`。

### 关键字参数（Keyword Argument）

带有`=`的是设定了默认值的参数，这就是关键字参数（Keyword Argument），缩写为`kwarg`。

在调用有 Keyword Argument 的函数时：
- 如若不提供这些参数，那么参数在执行时，启用的是它在定义时为那些 keyword Argument 设定的默认值。
- 如若提供这些参数的值，那么参数在执行时，启用的是接收到的相应值。

### 位置参数（Positional Argument）

位置参数，是“由位置决定其值的参数”。除关键字参数之外的参数，都是位置参数（Positional Argument），缩写为`arg`。

位置参数又可以分为如下几类：
- 可选位置参数（Optional Positional Argument）
- 可接收很多值的位置参数
